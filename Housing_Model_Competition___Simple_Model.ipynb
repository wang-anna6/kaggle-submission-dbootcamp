```python
TrainData = pd.read_csv('https://raw.githubusercontent.com/max56292/Data-Links/refs/heads/main/train.csv')
pd.set_option('display.max_columns', None)
TrainData.head(15)
```

```python
TrainData.columns
```

```python
TrainData[TrainData.columns[TrainData.isna().sum() > 0]].isna().sum()/len(TrainData)
TrainData["Alley"] = TrainData['Alley'].fillna = 0

TrainData["Alley"].head()
```

```python
X = TrainData[['OverallCond','GrLivArea','OverallQual','YearBuilt']]
y = TrainData['SalePrice']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=11)
```

```python
lr = LinearRegression()
```

```python
Model = lr.fit(X_train,y_train)
```

```python
test_rmse = np.sqrt(mean_squared_error(y_test, lr.predict(X_test)))
train_rmse = np.sqrt(mean_squared_error(y_train, lr.predict(X_train)))

print("test rmse is", test_rmse.round(2), "train rmse is", train_rmse.round(2))
```
